```python
import os, shutil, random

# Paths
images_dir = '/content/dataset/images'
labels_dir = '/content/dataset/labels'

train_img = '/content/dataset/train/images'
train_lbl = '/content/dataset/train/labels'
val_img = '/content/dataset/val/images'
val_lbl = '/content/dataset/val/labels'

# Create folders
for folder in [train_img, train_lbl, val_img, val_lbl]:
    os.makedirs(folder, exist_ok=True)

# Get all image files
images = os.listdir(images_dir)
random.shuffle(images)

# 80% train, 20% val
split = int(0.8 * len(images))
train_files = images[:split]
val_files = images[split:]

# Copy files
for img in train_files:
    shutil.copy(f'{images_dir}/{img}', f'{train_img}/{img}')
    lbl = os.path.splitext(img)[0] + '.txt'
    if os.path.exists(f'{labels_dir}/{lbl}'):
        shutil.copy(f'{labels_dir}/{lbl}', f'{train_lbl}/{lbl}')

for img in val_files:
    shutil.copy(f'{images_dir}/{img}', f'{val_img}/{img}')
    lbl = os.path.splitext(img)[0] + '.txt'
    if os.path.exists(f'{labels_dir}/{lbl}'):
        shutil.copy(f'{labels_dir}/{lbl}', f'{val_lbl}/{lbl}')

print(f'Train: {len(train_files)} images')
print(f'Val: {len(val_files)} images')
```

```python
import os

for root, dirs, files in os.walk('/content/dataset'):
    level = root.replace('/content/dataset', '').count(os.sep)
    indent = ' ' * 2 * level
    print(f'{indent}{os.path.basename(root)}/')
    if level < 2:
        for f in files[:3]:
            print(f'{indent}  {f}')
```

```python
yaml_content = """
path: /content/dataset
train: train/images
val: val/images

nc: 1
names: ['Krithika']
"""

with open('/content/dataset/data.yaml', 'w') as f:
    f.write(yaml_content)

print("data.yaml created successfully!")
```

```python
from ultralytics import YOLO

# Load YOLO26 model
model = YOLO('yolo26n.pt')

# Train
model.train(
    data='/content/dataset/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    name='custom_face_detection',
    patience=10
)
```

```python
import os, shutil, random, zipfile

# Re-extract zip
with zipfile.ZipFile('/content/dataset.zip', 'r') as zip_ref:
    zip_ref.extractall('/content/dataset')

# Create train/val folders
for folder in ['/content/dataset/train/images', '/content/dataset/train/labels',
               '/content/dataset/val/images', '/content/dataset/val/labels']:
    os.makedirs(folder, exist_ok=True)

# Split 80/20
images = os.listdir('/content/dataset/images')
random.shuffle(images)
split = int(0.8 * len(images))
train_files = images[:split]
val_files = images[split:]

for img in train_files:
    shutil.copy(f'/content/dataset/images/{img}', f'/content/dataset/train/images/{img}')
    lbl = os.path.splitext(img)[0] + '.txt'
    if os.path.exists(f'/content/dataset/labels/{lbl}'):
        shutil.copy(f'/content/dataset/labels/{lbl}', f'/content/dataset/train/labels/{lbl}')

for img in val_files:
    shutil.copy(f'/content/dataset/images/{img}', f'/content/dataset/val/images/{img}')
    lbl = os.path.splitext(img)[0] + '.txt'
    if os.path.exists(f'/content/dataset/labels/{lbl}'):
        shutil.copy(f'/content/dataset/labels/{lbl}', f'/content/dataset/val/labels/{lbl}')

print(f'Train: {len(train_files)} | Val: {len(val_files)}')
```

```python
from ultralytics import YOLO

model = YOLO('yolo26n.pt')

model.train(
    data='/content/dataset/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    name='dataset_face_detection',
    patience=10
)
```
